**Extended Data Figure 10 | CBE screen quality control, pipeline benchmarking, and hit identification.** a, The 90/10 skew ratio (counts of 90th percentile guide/counts of 10th percentile guide) for each indicated sample for CBE screen. b, Principal components analysis of the correlation matrix of barcode counts for replicates of indicated CBE samples. c, Average target base editing % (with bystander edits allowed, dark blue) and pure correct editing % (no bystander edits allowed, light blue) at the sensor locus. d, Histogram of target base editing rate (top, dark blue) and pure correct editing rate (bottom, light blue) at the sensor locus for all samples of CBE screen. e, Scatter plots of empirical −log10(FDR) versus median LFC (top) and MAGeCK −log10(FDR) versus MAGeCK LFC (bottom) for CBE gRNAs across all tissue contexts, with no sensor editing cutoff applied. Dashed red lines indicate FDR = 0.1. Points colored by hits classification: empirical framework only (blue), MAGeCK only (orange), called by both pipelines (red), or neither (gray). f, Venn diagrams comparing significant hits (FDR < 0.1) identified by MAGeCK versus the empirical framework for CBE gRNAs across all tissue contexts. MAGeCK substantially overestimates the number of significant hits relative to the empirical framework, with the greatest excess of false-positive calls observed in the meninges compartment. g, Waterfall plots of CBE gRNA enrichment scores across all tissue contexts (target base editing ≥ 20%), with the top 10 and bottom 10 gRNAs labeled per context. Points colored by COSMIC Cancer Gene Census annotation; point size scaled by FDR significance. h, Heatmap of pairwise Spearman correlation coefficients between per-replicate LFC z-scores for all CBE samples. Spleen and bone marrow replicates show moderate inter-replicate concordance, while meninges and in vitro samples exhibit lower correlations consistent with reduced selective pressure and variable engraftment.
